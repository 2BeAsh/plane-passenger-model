```python
import numpy as np
import plotter
import generate_data
```

```python
def load_data(fname):
    # Get the paths
    path_data = "./data/"
    passenger_path = path_data + fname + "_model.npz"
    times_path = path_data + f"{fname}_times_N{generate_data.N_repeat}.npz"
    
    # Load data
    passenger = np.load(passenger_path)["arr_0"]
    times = np.load(times_path)["arr_0"]
    return passenger, times

passenger_simple, times_simple = load_data(generate_data.fname_baseline)    
passenger_fullmove, times_fullmove = load_data(generate_data.fname_fullmove)
```

```python
# Animate simple passenger model
plotter.animate_passengers_scatter(passenger_simple, save=True, interval_ms=100)
```

```python
plotter.animate_passengers_scatter(passenger_fullmove, save=True, interval_ms=100, fname_add="fullmove_")
```

```python
Nbins = None
plotter.time_distribution(times_simple, Nbins)
plotter.time_distribution(times_fullmove, Nbins)
```
